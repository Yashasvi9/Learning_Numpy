INITIALIZING DIFFERENT TYPES OF ARRAYS

```python
#All 0s array
np.zeros((2,3))
```

```python
#All 1s array
np.ones((4,2,2), dtype = "int8")
```

```python
#Containing any other number instead of 0 and 1
np.full((2,2), 99, dtype = "int16")
```

```python
#Random decimal numbers (0-1) filled matrix 
np.random.rand(4,2)
```

```python
#Create a matrix of random elements
np.random.randint(7, size = (3,3))
```

```python
#Generate Identity matrix
np.identity(9)
```

```python
arr = np.array([[1,2,3]])
r1 = np.repeat(arr, 3, axis = 1)
print(r1)
```

```python
output = np.ones((5,5))
print(output)

z = np.zeros((3,3))
z[1,1] = 9
print(z)

output[1:4,1:4] = z
print(output)
```

### REORGANIZING ARRAY

```python
before = np.array([[1,2,3,4],[5,6,7,8]])
print(before)

after = before.reshape((4,2))
print(after)
```

```python
#Vertically stacking vectors

v1 = np.array([1,2,3,4])
v2 = np.array([5,6,7,8])

np.vstack([v1, v2])
```

```python
#Horizontal Stack
h1 = np.ones((2, 4))
h2 = np.zeros((2, 2))
np.hstack([h1, h2])
```
